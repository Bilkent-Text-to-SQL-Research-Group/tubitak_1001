Load the model

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_id = "ilhamiuturkkan/Generator-Qwen2.5"

# Tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_id, trust_remote_code=True)

# Model
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype="auto",
    trust_remote_code=True
)
```

Load the dataset

```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset("ilhamiuturkkan/dev-set-data-public")

print(len(dataset["train"]))
print(dataset.column_names)

filtered_dataset = dataset["train"].filter(lambda x: x['feedback_line'] == "Ground Truth")

# Print the number of rows in the filtered dataset
print(len(filtered_dataset))

# Print the column names
print(filtered_dataset.column_names)

dataset["train"] = filtered_dataset
```

Test each training data and store the results.

```python
from unsloth import FastLanguageModel
from unsloth.chat_templates import get_chat_template
import math
import json
import torch
import gc
from tqdm import tqdm

tokenizer = get_chat_template(
    tokenizer,
    chat_template = "qwen-2.5",
)
FastLanguageModel.for_inference(model) # Enable native 2x faster inference

objective = """Objective: You are an intermediate step generator. Your task is to determine which tables, columns, and values to use in order to generate an SQL query with respect to a given question. You must analyze the given query and given database schema. By analyzing both, you must extract the database entities related to the given question. You will have feedback based on your generated intermediate step.
Instructions:
1. Analyze the database schema and understand the structure of the database. Determine all the tables, their columns, and relations between tables.
2. Read the Question Carefully: Understand the primary focus and specific details of the question. Look for any named entities (such as organizations, locations, etc.), technical terms, and other phrases that encapsulate important aspects of the inquiry.
3. Analyze The Hint If It is Given: The hint is designed to direct attention toward certain elements relevant to answering the question. Extract any keywords, phrases, or named entities that could provide further clarity of direction in formulating an answer.
4. Generate an Intermediate Step: By analyzing the given database schema, question, and hint determine all the tables, those tables columns, and values to use in order to generate an SQL to retrieve the asked data.
"""

batch_size = 8
train_list = list(dataset["train"])
num_splits = 15
total      = len(train_list)
split_size = math.ceil(total / num_splits)

all_results = []

for part in range(num_splits):
    start_idx = part * split_size
    end_idx   = min(start_idx + split_size, total)
    subset    = train_list[start_idx:end_idx]
    part_results = []

    print(f"Processing split {part+1}/{num_splits}: items {start_idx}–{end_idx-1}")

    for i in tqdm(range(0, len(subset), batch_size)):
        batch = subset[i : i + batch_size]
        messages_batch = [
            [
                {
                    "role": "user",
                    "content": (
                        f"{objective}\n"
                        f"Database Schema: {e['db_schema']}\n"
                        f"Question: {e['question']}\n"
                        f"Hint: {e['evidence']}"
                    )
                }
            ]
            for e in batch
        ]

        # 1) Tokenize → could be Tensor or dict
        encoding = tokenizer.apply_chat_template(
            messages_batch,
            tokenize=True,
            add_generation_prompt=True,
            return_tensors="pt",
            padding=True,
        )

        # 2) Normalize it to a dict we can unpack
        if isinstance(encoding, torch.Tensor):
            encoding = {"input_ids": encoding}

        # 3) Move all tensors to GPU
        for k, v in encoding.items():
            if isinstance(v, torch.Tensor):
                encoding[k] = v.cuda()

        # 4) Inference
        with torch.no_grad():
            outputs = model.generate(
                **encoding,
                max_new_tokens=64,
                use_cache=True,
                temperature=1.5,
                min_p=0.1,
            )

        # 5) Decode & collect
        part_results.extend(tokenizer.batch_decode(outputs, skip_special_tokens=True))

        # 6) Cleanup
        del encoding, outputs
        gc.collect()
        torch.cuda.empty_cache()

    # save this part
    with open(f"results_part_{part+1}.json", "w", encoding="utf-8") as f:
        import json
        json.dump(part_results, f, ensure_ascii=False)

    all_results.extend(part_results)

print("Done — total outputs:", len(all_results))
```

```python
responses = []
for result in all_results:
  _, _, assistant_response = result.partition("assistant\n")
  responses.append(assistant_response.strip())
```

```python
import pandas as pd
from datasets import Dataset
ds = Dataset.from_dict({"response": responses})

ds.push_to_hub("ilhamiuturkkan/response-to-dev-generator-hub")
```

```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset("ilhamiuturkkan/dev-set-data-public")

print(len(dataset["train"]))
print(dataset.column_names)

filtered_dataset = dataset["train"].filter(lambda x: x['feedback_line'] == "Ground Truth")

# Print the number of rows in the filtered dataset
print(len(filtered_dataset))

# Print the column names
print(filtered_dataset.column_names)

dataset["train"] = filtered_dataset

dataset2 = load_dataset("ilhamiuturkkan/response-to-dev-generator-hub")

print(len(dataset2["train"]))
print(dataset2.column_names)
```

```python
import re
from collections import Counter      # (Counter is no longer used—can drop)

# ── 1. Regex that grabs the word after any "#table[digits]" tag ──────────────
TABLE_RE = re.compile(r"#table\d*\s+([A-Za-z0-9_]+)", re.IGNORECASE)

def extract_tables(text: str) -> set[str]:
    """
    Return the set of table names mentioned with a #table tag.
    Robust to:
      • '#table' lines that forget the name  -> ignored
      • '#table12   my_table'               -> captures 'my_table'
      • mixed case                          -> all lower-cased
    """
    return {m.group(1).lower() for m in TABLE_RE.finditer(text)}

# ── 2. Metrics loop — now tracks false positives for precision ───────────────
tp_total = fn_total = fp_total = exact_match = 0
total    = len(dataset["train"])

for gold_row, pred_row in zip(dataset["train"], dataset2["train"]):
    gold_tables = extract_tables(gold_row["inter_Step"])
    pred_tables = extract_tables(pred_row["response"])

    tp_total += len(gold_tables & pred_tables)      # true positives
    fn_total += len(gold_tables - pred_tables)      # false negatives
    fp_total += len(pred_tables - gold_tables)      # false positives

    # exact-match accuracy (all gold tables present in prediction)
    if gold_tables.issubset(pred_tables):
        exact_match += 1

recall    = tp_total / (tp_total + fn_total) if tp_total + fn_total else 0.0
precision = tp_total / (tp_total + fp_total) if tp_total + fp_total else 0.0
accuracy  = exact_match / total if total else 0.0

print(f"Recall:    {recall:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Accuracy:  {accuracy:.4f}")
```
